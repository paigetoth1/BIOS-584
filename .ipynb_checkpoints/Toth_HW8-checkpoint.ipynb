### Q2.1. Univariate comparison (3')

<font size='4'> 
    
- Suppose that we would like to examine the utility/effect of an intervention program for patients with PTSD.
- We measure PCL5 scores at completion (`pcl5week_score.completion`) and PCL5 score at 3-month follow-up (`pcl5month_score.3_month_follow_up`). Let's assume the first score is pre-intervention and the second score is post-intervention.
- Report the summary statistics for each variable including mean, std, median, Q1, and Q3.
- Note that each patient will receive such two PCL5 scores. Use a appropriate statistical test to perform the univariate comparison. Report the outputing statistic and p-value.
- Before you run the statistic test, determine the data type and check the missingness of two columns. In particular, report the number of NA values for each variable.

```python
# Write your own code

#pulling relevant columns and defining (pcl5week_score.completion and pcl5month_score.3_month_follow_up)
pre_scores_col = 'pcl5week_score.completion'
post_scores_col = 'pcl5month_score.3_month_follow_up'

#verifying columns
print(ptsd_df[[pre_scores_col, post_scores_col]].head(8))

#1: determining data type and checking missingness of pcl5week_score.completion and pcl5month_score.3_month_follow_up

print(ptsd_df[[pre_scores_col, post_scores_col]].dtypes) #data type verification
#both pre_scores and post_scores are float64

#summing and printing out the total number of missing values for the two columns

try:
    total_sum_na = (
        f"{pre_scores_col} has {number_na[pre_scores_col]} missing values. "
        f"{post_scores_col} has {number_na[post_scores_col]} missing values."
    )
    print(total_sum_na)
except NameError:
    print(f"Total NaNs in {pre_scores_col}: {ptsd_df[pre_scores_col].isna().sum()}")
    print(f"Total NaNs in {post_scores_col}: {ptsd_df[post_scores_col].isna().sum()}")


#1.1: cleaning missing values/converting errors to NaN

pre_scores_num = pd.to_numeric(ptsd_df[pre_scores_col], errors='coerce')
post_scores_num = pd.to_numeric(ptsd_df[post_scores_col], errors='coerce')

print(f"NaN values in pre_scores: {pre_scores_num.isna().sum()}")
print(f"NaN values in post_scores: {post_scores_num.isna().sum()}")


#1.2: creating a cleaned dataframe that drops rows with NaN values across both pre_scores and post_scores

dataset = pd.DataFrame({'pre':pre_scores_num, 'post':post_scores_num})
cleaned_dataset = dataset.dropna()

#extracting only the cleaned scores from pre and post columns
pre_scores_clean = cleaned_dataset['pre']
post_scores_clean = cleaned_dataset['post']


#2: reporting the summary statistics for each variable, including mean, std, median, Q1, and Q3

print("\n--- Summary Statistics (Pre-scores Cleaned) ---")
print(pre_scores_clean.describe()) 

print("\n--- Summary Statistics (Post-scores Cleaned) ---")
print(post_scores_clean.describe())

"""
3: Use an appropriate statistical test to perform a univariate comparison -> 
using paired t-test because PCL5 scores are continuous, and pre- and post-scores are related
"""
t_statistic, p_value = stats.ttest_rel(pre_scores_clean, post_scores_clean) #performing paired t-test

print(f"T-statistic: {t_statistic}") #printing t-test t statistic
print(f"P-value: {p_value}") #printing t-test p-value

alpha = 0.05 #setting significance level
if p_value < alpha:
    print("Reject the null hypothesis: There is a significant difference between the pre- and post-intervention PCL5 scores in this sample.")
else:
    print("Fail to reject the null hypothesis: There is no significant difference between the pre- and post-intervention PCL5 scores in this sample.")
```

### Q2.2. Multiple Linear Regression (7')

<font size='4'>

- Select columns specified in the following code chunk and create a subset dataset named `ptsd_sub_df`.
- Fit a linear regression to examine the association between `caps_intake` (continuous outcome) and the remaining covariates (as predictors) using `ptsd_sub_df`.
    - Note that all covariates ending with `_code` are categorical variables.
- Use the instruction here to write the formula for linear regression in Python.
    - https://www.statsmodels.org/stable/example_formulas.html
- Report the output page including R2, adjusted R2, and parameter estimates, SE, 95% confidence intervals, and p-values.
- Provide a brief interpretation for all significant predictors (p<0.05) excluding the intercept.
- Relevant label information includes:
    - `employment_code`: 1: Employed, 2: Unemployed, 3: Retired, 4: Disabled/Unable to work, 5: Student, 6: Other.
    - `rank_code`: 1. Enlisted, 2: Officer, 3: Other

```python
# The following column names are used for linear regression.
# Do not delete.
relevant_col_names = ['caps_intake', 'age_iop', 'gender_code', 'sexualorient_code', 'race_code', 'ethnicity_code', 
                      'education_code', 'employment_code',
                      'rank_code', 'branch_code', 'mdd_code', 'ctq_total_score', 'sexual_trauma', 'sud_code']
```

```python
# Write your own code

#1: select columns and create subset dataset ptsd_sub_df

ptsd_sub_df = ptsd_df[relevant_col_names].copy()

#2: fitting the model

mod = smf.ols(formula =
              'caps_intake ~ age_iop + ctq_total_score + sexual_trauma + '
                  'C(gender_code) + C(sexualorient_code) + C(race_code) + C(ethnicity_code)+ C(education_code) + C(employment_code) + C(rank_code) + C(branch_code) + C(mdd_code) + C(sud_code)',
                  data = ptsd_sub_df
             )

res = mod.fit()

print(res.summary())

#checking how many rows were dropped for being NaN
print("Rows before dropna:", ptsd_sub_df.shape[0])
ptsd_sub_df = ptsd_sub_df.dropna()
print("Rows after dropna:", ptsd_sub_df.shape[0])

"""
Reporting the output page including R2, adjusted R2, and parameter estimates,
SE, 95% confidence intervals, and p-values
"""
#reporting R2 and adjusted R2
print(f"R-Squared: {res.rsquared:.4f}")
print(f"Adjusted R-Squared: {res.rsquared_adj:.4f}")

#creating a results table
results_table = res.conf_int()
#parameter estimates, standard error, 95% CI, p-values, and t-values
ptsd_sub_df_results = pd.DataFrame({
    'Estimate': res.params,
    'StdErr': res.bse,
    'CI Lower': res.conf_int().iloc[:,0],   
    'CI Upper': res.conf_int().iloc[:,1],   
    'p-value': res.pvalues,
    't-value': res.tvalues
})
print(ptsd_sub_df_results)

#printing results in a table
ptsd_sub_df_results.style.background_gradient(cmap='Purples', subset=['p-value'])
```

# Write your interpretations below:

For the sake of simplicity, I have rounded all to the third decimal place.


In the multiple linear regression predicting CAPS intake, three predictors were significantly associated
with symptom severity after controlling for covariates (p < 0.05). Participants who were disabled or 
unable to work scored an average of 2.74 points higher on CAPS intake compared to those who were
employed (p = 0.029). Contrarily, participants with Officer ranks reported CAPS intake scores
approximately 3 points lower than those who were Enlisted (p = 0.022). Finally, higher scores on
the Childhood Trauma Questionnaire (CTQ) were associated with slightly increased CAPS intake, with
every one-point increase in CTQ corresponding to a 0.042-point increase in CAPS intake (p = 0.035).
This indicates a positive relationship between experiences of trauma in childhood and PTSD symptom severity.
