## Import Data

<font size='4'>

* Three lists named `study_id_ls`, `measure_date_ls`, and `duration_ls` have been initialized.
* Write a for loop to extract and append the values of `study_id`, `measure_date`, and `duration` to `study_id_ls`, `measure_date_ls`, and `duration_ls`, respectively.

```python
#creating empty lists for the study, measure date, and duration lists
study_id_ls = []
measure_date_ls = []
duration_ls = []

#for-loop to append data to the empty lists
for df in df_list:
    study_id_ls.extend(df.study_id.tolist())
    measure_date_ls.extend(df.measure_date.tolist())
    duration_ls.extend(df.duration.tolist())

print(study_id_ls)
print(measure_date_ls)
print(duration_ls)
```

<font size='4'>

* Create a new Pandas Dataframe combining three lists and name it `input_df`.

```python
#creating new dataframe combining the lists
input_df = pd.DataFrame({"study_id": study_id_ls,
       "measure_date": measure_date_ls,
       "duration": duration_ls
      })

#checking work
print(input_df)

#printing in readable table
input_df
```

<font size='4'>

* Show or print `input_df`

```python
#showing input_df as a table again
input_df
```

## Data Manipulation

<font size='4'>

* Reorder the dataframe by `measure_date` (earliest to latest) and then by `duration` (shortest to longest).

```python
#sorting measure_date from earliest date to latest date (ascending)
input_df = input_df.sort_values(by="measure_date", ascending = True)

#showing as a table
input_df

#soritng duration from the shortest duration to the longest duration (ascending)
input_df = input_df.sort_values(by="duration", ascending = True)

#showing as a table
input_df
```

<font size='4'>

* Create a new column called `relative_time` by calculating the time difference in days between each of the measure date and the earliest one.

```python
#creating relative_time column by calculating the difference in time for each date measure versus the earliest date

#importing datetime/date so that the dates will correctly substract and calculate
from datetime import date

input_df['measure_date'] = pd.to_datetime(input_df['measure_date'])

#pulling the earliest date recorded
earliest_date = input_df['measure_date'].min()
input_df['relative_time'] = (input_df['measure_date'] - earliest_date).dt.days

#checking work

input_df
```

<font size='4'>

* Reset the index column and remove the old index.

```python
input_df = input_df.reset_index(drop=True)
```
