### To permanently modify a variable, use the "global" command

```python
# 2.7.

# avoid using; may create confusion in code
def modify_x():
    global x
    x = x + 5

x = 1
print('x={} before modification'.format(x))
#now run the function will permanently increase x by 5
modify_x()
print('x={} after modify_x function'.format(x))
```
